```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
from langchain_core.documents import Document

doc1 = Document(
    page_content="The Indian Premier League (IPL) is a professional Twenty20 cricket league founded in 2008 by the BCCI. It is one of the most popular cricket leagues in the world.",
    metadata={
        "source": "wiki",
        "topic": "introduction",
        "year": 2008
    }
)

doc2 = Document(
    page_content="Mumbai Indians are the most successful team in IPL history, having won multiple titles. The team is led by Rohit Sharma and owned by Reliance Industries.",
    metadata={
        "team": "Mumbai Indians",
        "category": "team_info",
        "titles": 5
    }
)

doc3 = Document(
    page_content="Chennai Super Kings, captained by MS Dhoni, is known for its consistency and strong performances in the IPL playoffs.",
    metadata={
        "team": "Chennai Super Kings",
        "category": "team_info",
        "captain": "MS Dhoni"
    }
)

doc4 = Document(
    page_content="The IPL auction is held every year where franchises bid for players from around the world to build their squads.",
    metadata={
        "event": "IPL Auction",
        "category": "auction",
        "frequency": "annual"
    }
)

doc5 = Document(
    page_content="Virat Kohli is one of the highest run scorers in IPL history and has been a key player for Royal Challengers Bangalore.",
    metadata={
        "player": "Virat Kohli",
        "team": "RCB",
        "category": "player_stats"
    }
)
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="text-embedding-004"
)
```

```python
docs = [doc1,doc2,doc3,doc4,doc5]
```

```python
vectores = Chroma(
    embedding_function=embeddings,
    persist_directory='my_chroma_db',
    collection_name='sample'
)
```

```python
vectores.add_documents(docs)
```
