### Custome Tools

```python
from langchain_core.tools import tool

# step 1 - Create a function
def multiply(a,b):
    """Multiply two numbers"""
    # """Multiply two numbers""" is a docstring (documentation string)
    # used to describe what a function does.
    # Why it’s used :
        # readability 
        # for automatically describe tools
        # for automatically describe tools
    return a*b
```

```python
# step 2 - add type hint

def multiply(a:int,b:int)-> int :
    """Multiply two numbers"""
    return a*b
```

def multiply(a:int,b:int)-> int : this is type hunting process 
    Developers understand inputs
    Langchain understand expected arguments
    LLM generate correct tool calls.

    -> int

This means:

👉 “This function will return an integer”

Again, this is type hinting, not mandatory but very useful.

```python
@tool
def multiply(a:int,b:int)-> int :
    """Multiply two numbers"""
    return a*b
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

### Method 2 (StructuredTool)

```python
from langchain_community.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a : int = Field(required=True, description="The first number to add")
    b : int = Field(required=True, description="The second number to add")
```

```python
def multiply(a:int,b:int)-> int:
    return a* b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply,
    name="multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
multiply_tool.invoke({'a':3,'b':5})
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

### Method 3 - BaseTool class

```python
from langchain_community.tools import BaseTool
from pydantic import BaseModel, Field
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a : int = Field(required=True, description="The first number to add")
    b : int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description:str = "Multiply two numbers"
    args_schema : Type[BaseModel] = MultiplyInput

    def _run(self, a:int,b:int)->int:
        return a*b
```

```python
multiply_tool = MultiplyTool()
multiply_tool.invoke({'a':3,'b':5})
```
