```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = "Structured Query Language (SQL) is a standardized programming language used to manage and manipulate relational databases. Its history begins in 1970 when computer scientist Edgar F. Codd, working at IBM, introduced the concept of the relational database in his groundbreaking research paper titled “A Relational Model of Data for Large Shared Data Banks.” In this paper, Codd proposed organizing data into tables consisting of rows and columns, which could be linked through relationships rather than using the hierarchical or network database models that were common at the time."
```

### Documents splitters

```python
from langchain_community.document_loaders import TextLoader,PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
load_text = TextLoader("../data/sql_history.txt",encoding='utf-8')
text = load_text.load()
```

```python
load_pdf = PyPDFLoader("../data/JS1 ClassNotes.pdf")
pdf = load_pdf.load()
len(pdf)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=40)
```

```python
text_chunks = splitter.split_documents(text)
len(text_chunks)
```

```python
text_chunks
```

```python
pdf_chunks = splitter.split_documents(pdf)
pdf_chunks
```
