```python
from pydantic import BaseModel, Field
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name="mayur",age=23)
```

```python
user.name
```

```python
class Product(BaseModel):
    title:str = Field(max_length=10)
    des:str = Field(min_length=10)
    price:int = Field(gt=0,le=100)
```

```python
product = Product(title="Samsung", des="This is an amezing mobile.", price=23)
```
