```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a:int,b:int):
    """
    It will return sum of two numbers.
    Args:
        a : Number one
        b : Number two
    """
    return a+b

@tool
def multiply_number(a:int,b:int):
    """
    It will return product of two numbers.
    Args:
        a : Number one
        b : Number two
    """
    return a*b
```

```python
add_number.invoke({"a":3,"b":4})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
agent = create_agent(
    model = llm,
    tools = [add_number,multiply_number],
    system_prompt= "You are a math teacher, and always use tool for calculation."
)
```

```python
res = agent.invoke({"messages":[{"role":"user","content":"What is 3+4"}]})
```
