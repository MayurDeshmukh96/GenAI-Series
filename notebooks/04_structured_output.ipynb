```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

#### Discuss the problem first

```python
text = "hello, My name is Vikram. Email os  vikram@gmail and my age is 23."
```

```python
res = llm.invoke(f"please give me only name , email and age from this text :{text}")
print(res.content) 
# Here we get data in string but if i want output in structured output
```

```python
from pydantic import BaseModel, Field # Help to define data types
from typing import List

class ResponseStructure(BaseModel) :
    name: str = Field(description="Complete Name")
    email: str = Field(description="Email address")
    age: int = Field(description="Age")

structured_output = llm.with_structured_output(ResponseStructure)
```

```python
res = structured_output.invoke(f"please give me only name , email and age from this text :{text}")
```

```python
print(res.name)
print(res.email)
print(res.age)
```

```python
class movies(BaseModel):
    title:str = Field(description="Movie Title")
    year:int = Field(description="Movie Released Year")

class AllMovies(BaseModel):
    movies:List[movies]

movies_llm = llm.with_structured_output(AllMovies)
```

```python
res = movies_llm.invoke("Give me three sci-fi movie")
```
