#### connect with google gemini model

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

```python
res = llm.invoke("What is genai? explain in short")
```

```python
print(res.content)
```

#### Static prompts

```python
propmts = [
    ("system","Act as python developer."),
    ("user","How to perform sorting.")
]

res = llm.invoke(propmts)
print(res.content)
```
