### Conditional workflow

```python
# Conditional worklflow looking like parallel workflow it also have 
# mupltiple branches but we can't enter into all branches at time like 
# parallel workflow. 

# We enter into brach based on condition.

# COnditional workflow is like a "if-else" statement 
# We can add multiple branches in conditional worflow.
```

#### Non llm flow for understanding

```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Literal
from rich import print
```

```python
class Quadrent(TypedDict):
    a : int
    b : int
    c : int

    equation : str
    discriminant : float
    result : str
    
```

```python
def show_equation(state:Quadrent):

    equation = f'{state['a']}x2{state['b']}x{state['c']}'
    
    return {'equation':equation}

def calculate_discriminant(state:Quadrent):

    discriminant = state['b']**2 - (4*state['a']*state['c'])

    return {'discriminant':discriminant}
```

```python
def real_roots(state:Quadrent):

    root1 = (-state['b']+state['discriminant']**0.5)/(2*state['a'])
    root2 = (-state['b']-state['discriminant']**0.5)/(2*state['a'])

    result = f'The roots are {root1} and {root2}'

    return{'result':result}


def repeated_roots(state:Quadrent):
    root = (-state['b'])/(2*state['a'])

    result = f'Only repeating roots is {root}'

    return {'result':result}



def no_real_roots(state:Quadrent):

    result = f"No real roots"

    return{'result':result}



def check_condition(state:Quadrent) -> Literal['real_roots','repeated_roots','no_real_roots']:

    if state['discriminant'] > 0:
        return 'real_roots'
    elif state['discriminant']==0:
        return 'repeated_roots'
    else:
        return "no_real_roots"
```

```python
graph = StateGraph(Quadrent)

graph.add_node('show_equation',show_equation)
graph.add_node('calculate_discriminant',calculate_discriminant)
graph.add_node('real_roots',real_roots)
graph.add_node('no_real_roots',no_real_roots)
graph.add_node('repeated_roots',repeated_roots)


graph.add_edge(START,'show_equation')
graph.add_edge('show_equation','calculate_discriminant')

graph.add_conditional_edges('calculate_discriminant',check_condition)
graph.add_edge('calculate_discriminant',END)


workflow = graph.compile()
```

```python
initial = {'a':4, 'b':2 , 'c':4}

print(workflow.invoke(initial))
```
