```python
from langgraph.graph import StateGraph,START,END
from dotenv import load_dotenv
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
# Define State

class blogstate(TypedDict):

    topic : str
    outline : str
    blog : str
```

```python
# Define nodes

# globally model declaration
model = ChatGoogleGenerativeAI(model='gemini-2.5-flash-lite')

# -------------------Outline node----------------
def generate_outline(state:blogstate)-> blogstate:

    # Take topic
    topic = state['topic']

    # Generate Outline prompt
    prompt = f'Generate outline form this {topic} for blog'

    # generate outline
    state['outline'] = model.invoke(prompt)

    return state

# -------------------blog node----------------

def generate_blog(state:blogstate)-> blogstate:

    # Take outline 
    outline = state['outline']

    # Generate prompt
    prompt = f'Generate detailed blog based on this outline {outline}'

    # Generate Blog
    state['blog'] = model.invoke(prompt)

    return state
```

```python
# Define graph
graph = StateGraph(blogstate)

# Add nodes 
graph.add_node('generate_outline',generate_outline)
graph.add_node('generate_blog',generate_blog)

# Add edges
graph.add_edge(START,'generate_outline')
graph.add_edge('generate_outline','generate_blog')
graph.add_edge('generate_blog',END)

# Compile

blog_llm = graph.compile()
```

```python
initial_state = {'topic':'Increasing heat in india'}
final_state = blog_llm.invoke(initial_state)
print(final_state['blog'])
```
