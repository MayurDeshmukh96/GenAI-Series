```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain_community.utilities import GoogleSerperAPIWrapper
# from langchainhub import hub                                        # ← direct from langchainhub
from langchain.agents import create_agent
from rich import print
```

```python
model = ChatGoogleGenerativeAI(model='gemini-2.5-flash-lite')
```

```python
@tool
def search_on_google(query:str)-> str:
    """This search on google based on query"""
    search = GoogleSerperAPIWrapper()
    ans = search.run(query)
    return ans
```

```python
print(search_on_google.invoke({'query':"today ipl match"}))
```

```python
agent = create_agent(
    model=model,
    tools=['search_on_google'],
)
```
