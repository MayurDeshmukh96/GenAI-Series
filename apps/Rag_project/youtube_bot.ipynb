```python
retriver = vectore_store.as_retriever(search_type='similarity',search_kwargs={'k':4},)
```

```python
retriver.invoke('what cover in google conversation ')
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

```python
prompt = PromptTemplate(
    template="""Yor are a helpfull assistant.
    Don't give me any other answer if answer is not present in context.
    If answer is not present in context just say I don;t know.

    {context}

    Question : {question}
    """,
    input_variables=['context','question']
)
```

```python
question = "What did Pichai say AI models could potentially do to existing software?"
retrived_docs = retriver.invoke(question)
retrived_docs
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrived_docs)
context_text
```

```python
final_prompt = prompt.invoke({'context':context_text,"question":question})
final_prompt
```

```python
answer = llm.invoke(final_prompt)
print(answer)
```

```python
from langchain_core.runnables import RunnableLambda, RunnableParallel, RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_doc(retrived_docs):
    context_text = '\n\n'.join(doc.page_content for doc in retrived_docs)
    return context_text
```

```python
parallel_chain = RunnableParallel({
    'context' : retriver | RunnableLambda(format_doc),
    'question':RunnablePassthrough()
}
)
```

```python
parallel_chain.invoke('What did Pichai say AI models could potentially do to existing software?')
```

```python
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke('What did Pichai say AI models could potentially do to existing software?')
```
